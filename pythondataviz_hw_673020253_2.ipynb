## ให้สร้างฟังก์ชั่นคำนวณเวลาเป็นวินาทีของเวลาใดๆ และ print ออกมาให้สวยงาม


````python
def counting_seconds(time):
    t = time.split(':')
    h = int(t[0])
    m = int(t[1])
    se = int(t[2])
    total_second = (h * 3600) + (m * 60) + se
    print(f"เวลา {time} ชั่วโมง คิดเป็น {total_second:,} วินาที")
````

## ให้ สร้าง list ของ เวลา ```['12:30:15','13:41:07',....]``` แล้ววนลูปเรียกฟังก์ชั่นคำนวณเวลาเป็น วินาที

**commit ว่า HW3**

````python
time_list = [
    "12:30:15",
    "13:41:07",
    "12:53:15",
    "00:59:25",
    "11:11:11",
    "16:06:09"]
````

````python
print(f'<-------แปลงเวลา(ชั่วโมง) เป็นวินาที------->')
for i in time_list:
    counting_seconds(i)
````

## HW
#### เขียน function ตัดเกรด โดยที่ input เป็นคะแนน(0-100) , output เป็นเกรด (F-A)
#### วนลูปตัดเกรด input = [1,50,65,90,101,-5,49]

##### กำหนด
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ  0  แต่น้อยกว่า 50 จะได้เกรด F
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 50 แต่น้อยกว่า 55 จะได้เกรด D
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 55 แต่น้อยกว่า 60 จะได้เกรด D+
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 60 แต่น้อยกว่า 65 จะได้เกรด C
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 65 แต่น้อยกว่า 70 จะได้เกรด C+
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 70 แต่น้อยกว่า 75 จะได้เกรด B
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 75 แต่น้อยกว่า 80 จะได้เกรด B+
      ถ้าคะแนนอยู่ระหว่างมากกว่าหรือเท่ากับ 80  จะได้เกรด A
      แต่ว่าค่าคะแนนจะต้องไม่ต่ำกว่า 0 และมากกว่า 100 ไม่งั้นจะเกิด error

````python
def grade(score):
    if score > 100:
        return "Error"
    elif score < 0:
        return "Error"
    elif score >= 80:
        return "A"
    elif score >=75:
        return "B+"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C+"
    elif score >= 60:
        return "C"
    elif score >= 55:
        return "D+"
    elif score >= 50:
        return "D"
    else:
        return "F"
````

````python
score_list = [1,50,65,90,101,-5,49]
````

````python
for i in score_list:
    if 0 <= i <= 100:
        score_grade = grade(i)
        print(f"คะแนนคุณคือ {i} คุณได้เกรด {score_grade}")
    else:
        print(f"คะแนนคุณคือ {i} ไม่ถูกต้อง")
````
